```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split

df_maternal = pd.read_csv("../data/raw/maternal_health_risk.csv")
```

```python
print("\n   MATERNAL HEALTH RISK ÖN İŞLEME")

print("\n------------------------------------------------------------------------------\n")

print("Orijinal Veri Seti Boyutu:", df_maternal.shape)

print("\n------------------------------------------------------------------------------\n")

maternal = df_maternal.copy()
ham_veri_boyutu = len(maternal)

print("Eksik hedef (RiskLevel) sayısı:", maternal["RiskLevel"].isnull().sum())

print("\n------------------------------------------------------------------------------\n")

maternal["RiskLevel"] = maternal["RiskLevel"].map({"low risk": 0, "mid risk": 1, "high risk": 2})
print("Dönüştürülmüş Hedef Değişken Dağılımı:\n")
print(maternal['RiskLevel'].value_counts().sort_index())

print("\n------------------------------------------------------------------------------\n")

onceki = len(maternal)
maternal = maternal.drop_duplicates().reset_index(drop=True)
print("Silinen tekrar sayısı:", onceki - len(maternal), "| Kalan:", len(maternal))
print("Tekrarlanan değerler silindikten sonraki veri boyutu:", len(maternal))

print("\n------------------------------------------------------------------------------\n")

print("Eksik Değer Sayısı:")
print(df_maternal.isnull().sum())

print("\n------------------------------------------------------------------------------\n")

onceki = len(maternal)
maternal = maternal[maternal["HeartRate"] >= 30].reset_index(drop=True)
print("Silinen hatalı nabız sayısı:", onceki - len(maternal), "| Kalan:", len(maternal))
print("Hatalı nabız değerleri silindikten sonraki veri boyutu:", len(maternal))

print("\n------------------------------------------------------------------------------\n")

print("Sızıntı Kontrolü:\n")
print(maternal.corr()["RiskLevel"].drop("RiskLevel").abs().sort_values(ascending=False))

print("\n------------------------------------------------------------------------------\n")

X = maternal.drop(columns="RiskLevel")
y = maternal["RiskLevel"]

print("Özellikler (X) Boyutu:", X.shape)
print("Hedef (y) Boyutu:", y.shape)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("\n------------------------------------------------------------------------------\n")

print(f"Maternal Health Risk: {ham_veri_boyutu} -> {len(maternal)} satır | train {len(X_train)}, test {len(X_test)}")

print("\n------------------------------------------------------------------------------")

print("\nTrain sınıf dağılımı:\n", y_train.value_counts().sort_index())
print("\n--------------------------------")
print("\nTest sınıf dağılımı:\n", y_test.value_counts().sort_index())
print("\n--------------------------------")
      
print("\nTrain sınıf oranları (%):\n", (y_train.value_counts(normalize=True).sort_index() * 100).round(1))
print("\n--------------------------------")
print("\nTest sınıf oranları (%):\n", (y_test.value_counts(normalize=True).sort_index() * 100).round(1))
print("\n--------------------------------\n")

X_train.assign(RiskLevel=y_train).to_csv("../data/processed/maternal_train.csv", index=False)
X_test.assign(RiskLevel=y_test).to_csv("../data/processed/maternal_test.csv", index=False)
```
