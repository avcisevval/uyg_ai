```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split

df_gdm = pd.read_csv("../data/raw/gestational_diabetes.csv")
```

```python
print("\n   GDM ÖN İŞLEME")

print("\n------------------------------------------------------------------------------\n")

print("Orijinal Veri Seti Boyutu:", df_gdm.shape)

print("\n------------------------------------------------------------------------------\n")

gdm = df_gdm.copy()
ham_veri_boyutu = len(gdm)

print("Eksik hedef (Prediction) sayısı:", gdm["Prediction"].isnull().sum())

gdm.columns = gdm.columns.str.strip() #Sütun adlarındaki baş/son boşlukları temizleme

print("\n------------------------------------------------------------------------------\n")

print("Hedef değerleri:", gdm["Prediction"].unique())

print("\n------------------------------------------------------------------------------\n")

onceki = len(gdm)
gdm = gdm.drop_duplicates().reset_index(drop=True)
print("Silinen tekrar sayısı:", onceki - len(gdm), "| Kalan:", len(gdm))
print("Tekrarlanan değerler silindikten sonraki veri boyutu:", len(gdm))

print("\n------------------------------------------------------------------------------\n")

print("Eksik Değer Sayısı:")
print(df_gdm.isnull().sum())

print("\n------------------------------------------------------------------------------\n")

print("Metin (object) sütun sayısı:", (gdm.dtypes == "object").sum())

print("\n------------------------------------------------------------------------------\n")

print("Sızıntı Kontrolü:\n")
print(gdm.corr()["Prediction"].drop("Prediction").abs().sort_values(ascending=False).head(6))

print("\n------------------------------------------------------------------------------\n")

X = gdm.drop(columns="Prediction")
y = gdm["Prediction"]

print("Özellikler (X) Boyutu:", X.shape)
print("Hedef (y) Boyutu:", y.shape)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("\n------------------------------------------------------------------------------\n")

print(f"GDM: {ham_veri_boyutu} -> {len(gdm)} satır | train {len(X_train)}, test {len(X_test)}")

print("\n------------------------------------------------------------------------------\n")

print("Train sınıf dağılımı:\n", y_train.value_counts().sort_index())
print("\n--------------------------------\n")
print("Test sınıf dağılımı:\n", y_test.value_counts().sort_index())
print("\n--------------------------------")

print("\nTrain sınıf oranları (%):\n", (y_train.value_counts(normalize=True).sort_index() * 100).round(1))
print("\n--------------------------------")
print("\nTest sınıf oranları (%):\n", (y_test.value_counts(normalize=True).sort_index() * 100).round(1))
print("\n--------------------------------\n")

X_train.assign(Prediction=y_train).to_csv("../data/processed/gdm_train.csv", index=False)
X_test.assign(Prediction=y_test).to_csv("../data/processed/gdm_test.csv", index=False)
```
